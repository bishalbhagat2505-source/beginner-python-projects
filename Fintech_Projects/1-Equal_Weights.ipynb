```python
# importing necessary libraries
import pandas as pd
import numpy as np
import math
import yfinance as yf
from transformers.time_series_utils import LambdaLayer
```

```python
#loading nifty data set
tickers=pd.read_csv(r"dataset/ind_nifty50list.csv")
tickers.head()
```

```python
#removing un required columns and adding a ticker column for yfinance
tickers['Symbol']=tickers['Symbol']+'.NS'
tickers=tickers.drop(['Industry','Series','ISIN Code'],axis=1)
```

```python
tickers.head()
```

#

```python
#function for loading the market cap and sort the tickers with respect to marketcap

def fetch_market_cap(tickers_list):
    data=yf.download(tickers_list,period='1d',group_by='ticker',auto_adjust=False)
    stocks_data=[]
    for ticker in tickers_list:
        latest_price=data[ticker]['Close'].iloc[-1]
        market_cap= yf.Ticker(ticker).info.get("marketCap",'N/A')
        stocks_data.append({
           "Ticker":ticker,
           "Market Cap": market_cap,
           "Latest Price": latest_price,
       })
    stock_df=pd.DataFrame(stocks_data)
    return stock_df

tickers_list=tickers['Symbol'].values.tolist()
df=fetch_market_cap(tickers_list)
df.head()
```

```python
df=df.sort_values("Market Cap",ascending=False)
df
```

```python
#taking only top 10 stocks as per Market Cap
df=df.head(10)
df.reset_index(inplace=True,drop=True)
df
```

```python
portfolio_size=int(input("Enter the amount to invest : "))
position_size=portfolio_size/len(df.index)
```

```python
df['Number of shares to buy']=df['Latest Price'].apply(lambda price: math.floor(
    position_size/price
))
```
